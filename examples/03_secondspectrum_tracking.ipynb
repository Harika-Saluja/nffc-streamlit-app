# 03 · SecondSpectrum tracking

25 fps player + ball tracking for Premier League games.

**Heads up:** each game's `*_Data.jsonl` is ~400 MB. We download one game
locally first, then read it — much faster than streaming repeatedly.


```python
import nffc_data as nffc
from nffc_data import ssio
```

## Find a game
Pick a season (`202324`, `202425`, partial `202526`) and list games:


```python
games = nffc.list_bucket(prefix="SecondSpectrum/202425", depth=1)
list(games)[:5]
```

## Resolve the file keys for a game


```python
# game id with or without leading "g"
keys = ssio.game_files(season="202425", game_id="g2561895")
keys
```

## Read the metadata (small)


```python
meta = ssio.read_metadata(keys["metadata"])
print(meta["description"], "-", meta.get("fps"), "fps")
meta["homePlayers"][:2]
```

## Download the tracking once, then read locally


```python
local = nffc.download_file(keys["data"], "data/tracking/g2561895_Data.jsonl")
frames = ssio.read_tracking_local("data/tracking/", "g2561895_Data.jsonl")
print(frames.shape)
frames[["period", "frameIdx", "gameClock", "lastTouch"]].head()
```
