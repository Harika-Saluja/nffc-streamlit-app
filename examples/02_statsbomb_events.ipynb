# 02 · StatsBomb events

Matches, events and lineups for the top-5 leagues (2022-23, 2023-24, 2024-25).
See `docs/data_dictionary.md` and `docs/providers/Statsbomb/`.


```python
from nffc_data.statsbomb import StatsBombLoader

# Optional: cache_dir keeps a local copy of anything you read.
sb = StatsBombLoader("Premier League", "2023-2024", cache_dir="data/sb_cache")
```

## Matches


```python
matches = sb.load_matches()
print(matches.shape)
matches[["match_id", "match_date", "home_team", "away_team", "match_week"]].head()
```

## Events for one match


```python
mid = matches.iloc[0]["match_id"]
events = sb.load_events_for_match(mid)
print(events.shape)
events[["minute", "second", "type", "team", "player"]].head(15)
```

## Lineups for one match


```python
lineups = sb.load_lineups_for_match(mid)
list(lineups)[:2] if isinstance(lineups, dict) else lineups[:1]
```

## Filter to a team
`StatsBombLoader` can filter matches by team or matchweek:


```python
forest = sb.load_matches(team="Nottingham Forest")
forest.shape
```
