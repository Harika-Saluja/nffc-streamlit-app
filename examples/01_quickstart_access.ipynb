## Saving your work (your personal folder)

⚠️ Shared bucket: writes are real and have **no undo**. The `upload_*` helpers
default to `students/<you>/` (from `WASABI_USER`) — keep your work there and
never write over the shared datasets.


```python
import pandas as pd
my_df = pd.DataFrame({"player": ["example"], "metric": [1.0]})
key = nffc.upload_parquet(my_df, "scratch/first_save.parquet")
print("saved to:", key)            # -> students/<you>/scratch/first_save.parquet
nffc.load_parquet(key).head()      # read your own file back
```
