# 04 · Catapult GPS (session level)

10 Hz GPS per **training session** for NFFC athletes, with real athlete names.
One parquet per session, at `Catapult/activity/season=.../date=.../`.


```python
import nffc_data as nffc
```

## Browse available sessions


```python
nffc.list_bucket(prefix="Catapult/activity", depth=2)
```

## Read one session


```python
# replace with a real key from the listing above
key = nffc.ls("Catapult/activity")[0]
key
```

```python
gps = nffc.load_parquet(key)
print(gps.shape)
gps[["athlete_name", "activity_name"]].drop_duplicates().head()
```

## Per-athlete summary


```python
gps.groupby("athlete_name").size().sort_values(ascending=False).head()
```
