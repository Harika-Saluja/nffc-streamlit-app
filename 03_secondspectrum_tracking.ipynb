# 03 · SecondSpectrum tracking

25 fps player + ball tracking for Premier League games.

**Heads up:** each game's `*_Data.jsonl` is ~400 MB. We pick a real game,
read its small metadata, then download one game and read it locally.


```python
import nffc_data as nffc
from nffc_data import ssio
```

## Find a game
Seasons available: `202324`, `202425`, `202526`. List the games in one:


```python
season = "202425"
games = [k.rstrip("/").split("/")[-1] for k in nffc.ls(f"SecondSpectrum/{season}")]
print(len(games), "games; e.g.", games[:3])
game = games[0]
```

## Read the metadata (small)


```python
keys = ssio.game_files(season, game)
meta = ssio.read_metadata(keys["metadata"])
print(meta["description"], "-", meta.get("fps"), "fps")
meta["homePlayers"][:2]
```

## Download the tracking once (~400 MB), then read locally


```python
local = nffc.download_file(keys["data"], f"data/tracking/{game}_Data.jsonl")
frames = ssio.read_tracking_local("data/tracking/", f"{game}_Data.jsonl")
print(frames.shape)
frames[["period", "frameIdx", "gameClock", "lastTouch"]].head()
```

Each frame's `homePlayers`/`awayPlayers` carry `optaId` — join to the
metadata players for names/positions, and to other datasets via the mapping.
